Importing Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Importing Dataset- Table 1


```python
df=pd.read_csv('india_fuel_prices.csv')
df.head()
```

Data Analysis


```python
df= df.rename(columns={'Unnamed: 0': 'City'})
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['City'].nunique()
```

```python
#Removing rupees/L sign
df['diesel'] = df['diesel'].str[:-4]
df['petrol'] = df['petrol'].str[:-4]
df.head()
```

```python
df['diesel'] = df['diesel'].astype('float')
df['petrol'] = df['petrol'].astype('float')
df.head()
```

```python
df
```

Max price


```python
Max_diesel = df.sort_values(by = ['diesel'], ascending=False)
Max_petrol = df.sort_values(by = ['petrol'], ascending=False)
print("City with highest diesel price:", Max_diesel.iloc[0]['City'], "with price:", Max_diesel.iloc[0]['diesel'])
print("City with highest petrol price:", Max_petrol.iloc[0]['City'], "with price:", Max_petrol.iloc[0]['petrol'])
```
